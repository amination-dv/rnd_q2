```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import xarray as xr
from PIL import Image
```

```python
def get_tubeview(path: Path, shape: tuple[int, int] = None) -> np.ndarray:
    da = xr.load_dataarray(path)
    da /= da.max()
    da = da.fillna(1.0)
    da = (255 * da.to_numpy()).round().astype(np.uint8)
    if shape is None:
        return np.array(Image.fromarray(da, mode="L"))
    return np.array(Image.fromarray(da, mode="L").resize(shape, Image.LANCZOS))
```

```python
input_dir = Path("dataset")
output_dir = Path("dataset_images")
output_dir.mkdir(exist_ok=True)

nc_files = list(input_dir.glob("**/*.nc"))

for nc_file in nc_files:
    relative_path = nc_file.relative_to(input_dir).with_suffix(".png")
    output_path = output_dir / relative_path
    output_path.parent.mkdir(parents=True, exist_ok=True)
    np_array = get_tubeview(nc_file).T
    img = Image.fromarray(np_array, mode="L")
    img.save(output_path)
```
