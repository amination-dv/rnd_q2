```python
import matplotlib.pyplot as plt
import numpy as np
plt.style.use("dark_background")

# Data for ILIT0011
categories = ['Not Sized\nApproved', 'Poor Data\nQuality', 'False\nPositives', 'Approved']

# Arm-encoder v3 data
arm_not_sized = {'Other': 65, 'Weld': 1391, 'Metal Loss': 302, 'Deformation': 6949}  # Total: 8747
arm_poor_quality = 70
arm_false_positives = 6760
arm_approved = {'Other': 0, 'Weld': 1, 'Metal Loss': 0, 'Deformation': 716}  # Total: 717

# Ultrasound v1 data
us_not_sized = {'Other': 9187, 'Weld': 990, 'Metal Loss': 1164, 'Deformation': 2530}  # Total: 13861
us_poor_quality = 6
us_false_positives = 1895
us_approved = {'Other': 2, 'Weld': 1, 'Metal Loss': 1, 'Deformation': 726}  # Total: 730

# Colors for stacked bars
colors = {'Deformation': '#2ecc71', 'Metal Loss': '#e74c3c', 'Weld': '#3498db', 'Other': '#9b59b6'}

fig, axes = plt.subplots(1, 2, figsize=(16, 8))

# Plot 1: Arm-encoder v3
ax1 = axes[0]
x = np.arange(len(categories))
width = 0.6

# For stacked categories
bottom_not_sized = 0
bottom_approved = 0

# Not Sized - Approved (stacked)
for cat in ['Deformation', 'Metal Loss', 'Weld', 'Other']:
    val = arm_not_sized[cat]
    ax1.bar(x[0], val, width, bottom=bottom_not_sized, color=colors[cat])
    if val > 200:
        ax1.text(x[0], bottom_not_sized + val/2, f'{val}', ha='center', va='center', fontsize=9, fontweight='bold', color='white')
    bottom_not_sized += val

# Poor Data Quality (single)
ax1.bar(x[1], arm_poor_quality, width, color='#7f8c8d')
ax1.text(x[1], arm_poor_quality/2, f'{arm_poor_quality}', ha='center', va='center', fontsize=10, fontweight='bold', color='white')

# False Positives (single)
ax1.bar(x[2], arm_false_positives, width, color='#e67e22')
ax1.text(x[2], arm_false_positives/2, f'{arm_false_positives}', ha='center', va='center', fontsize=10, fontweight='bold', color='white')

# Approved (stacked)
for cat in ['Deformation', 'Metal Loss', 'Weld', 'Other']:
    val = arm_approved[cat]
    ax1.bar(x[3], val, width, bottom=bottom_approved, color=colors[cat])
    if val > 50:
        ax1.text(x[3], bottom_approved + val/2, f'{val}', ha='center', va='center', fontsize=9, fontweight='bold', color='white')
    bottom_approved += val

# Add totals on top
ax1.text(x[0], sum(arm_not_sized.values()) + 200, f'Total: {sum(arm_not_sized.values())}', ha='center', va='bottom', fontsize=10, fontweight='bold')
ax1.text(x[3], sum(arm_approved.values()) + 200, f'Total: {sum(arm_approved.values())}', ha='center', va='bottom', fontsize=10, fontweight='bold')

ax1.set_xticks(x)
ax1.set_xticklabels(categories, fontsize=11)
ax1.set_ylabel('Count', fontsize=12)
ax1.set_title('ILIT0011 - Arm-encoder v3', fontsize=14, fontweight='bold')
# Create legend patches for all categories
import matplotlib.patches as mpatches
legend_patches = [
    mpatches.Patch(color='#2ecc71', label='Deformation'),
    mpatches.Patch(color='#e74c3c', label='Metal Loss'),
    mpatches.Patch(color='#3498db', label='Weld'),
    mpatches.Patch(color='#9b59b6', label='Other'),
    mpatches.Patch(color='#7f8c8d', label='Poor Data Quality'),
    mpatches.Patch(color='#e67e22', label='False Positives'),
]
ax1.legend(handles=legend_patches, loc='upper right', fontsize=9)
ax1.grid(axis='y', alpha=0.3)

# Plot 2: Ultrasound v1
ax2 = axes[1]

# For stacked categories
bottom_not_sized = 0
bottom_approved = 0

# Not Sized - Approved (stacked)
for cat in ['Deformation', 'Metal Loss', 'Weld', 'Other']:
    val = us_not_sized[cat]
    ax2.bar(x[0], val, width, bottom=bottom_not_sized, color=colors[cat])
    if val > 200:
        ax2.text(x[0], bottom_not_sized + val/2, f'{val}', ha='center', va='center', fontsize=9, fontweight='bold', color='white')
    bottom_not_sized += val

# Poor Data Quality (single)
ax2.bar(x[1], us_poor_quality, width, color='#7f8c8d')
ax2.text(x[1], us_poor_quality + 100, f'{us_poor_quality}', ha='center', va='bottom', fontsize=10, fontweight='bold')

# False Positives (single)
ax2.bar(x[2], us_false_positives, width, color='#e67e22')
ax2.text(x[2], us_false_positives/2, f'{us_false_positives}', ha='center', va='center', fontsize=10, fontweight='bold', color='white')

# Approved (stacked)
for cat in ['Deformation', 'Metal Loss', 'Weld', 'Other']:
    val = us_approved[cat]
    ax2.bar(x[3], val, width, bottom=bottom_approved, color=colors[cat])
    if val > 50:
        ax2.text(x[3], bottom_approved + val/2, f'{val}', ha='center', va='center', fontsize=9, fontweight='bold', color='white')
    bottom_approved += val

# Add totals on top
ax2.text(x[0], sum(us_not_sized.values()) + 200, f'Total: {sum(us_not_sized.values())}', ha='center', va='bottom', fontsize=10, fontweight='bold')
ax2.text(x[3], sum(us_approved.values()) + 200, f'Total: {sum(us_approved.values())}', ha='center', va='bottom', fontsize=10, fontweight='bold')

ax2.set_xticks(x)
ax2.set_xticklabels(categories, fontsize=11)
ax2.set_ylabel('Count', fontsize=12)
ax2.set_title('ILIT0010 - Ultrasound v1', fontsize=14, fontweight='bold')
ax2.legend(handles=legend_patches, loc='upper right', fontsize=9)
ax2.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('ILIT0011_comparison.png', dpi=150, bbox_inches='tight', facecolor='white')
plt.show()

print("Plot saved as 'ILIT0011_comparison.png'")
```
